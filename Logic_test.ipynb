```python
p = torch.tensor([[1,2,3],[1,2,3], [1,2,3]]) # (3,3)
q = torch.tensor([[0,2,3],[1,2,3], [1,2,3]]) # (3,3)
batch_size = p.shape[0] # batch_size : int = 2(行数をあらわす。)
print(batch_size) # expected to get 3, why 2? -> 
# Compute only the lower triangular elements if p == q
tril_i, tril_j = torch.tril_indices(batch_size, batch_size, offset=-1)
print(f'tril_i:{tril_i}, tril_j: {tril_j}')
hellinger_tril = torch.sqrt(
            0.5 * torch.sum((torch.sqrt(p[tril_i]) - torch.sqrt(q[tril_j])) ** 2, dim=1)
)
print(f'hellinger_tril: {hellinger_tril}')
```
